# 01 – Разведочный анализ данных (EDA)

В этом ноутбуке:

* загружаем исходный датасет по риску сердечно-сосудистых заболеваний;
* выполняем базовый анализ распределений и пропусков;
* делаем простую очистку данных;
* сохраняем очищенный датасет в `data/processed/cvd_clean.csv`.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Пути к данным
PROJECT_ROOT = Path("..").resolve()
RAW_DATA_PATH = PROJECT_ROOT / "data" / "raw" / "cvd_raw.csv"
PROCESSED_DATA_PATH = PROJECT_ROOT / "data" / "processed" / "cvd_clean.csv"

RAW_DATA_PATH, PROCESSED_DATA_PATH
```

```
# Загрузка исходного датасета
df = pd.read_csv(RAW_DATA_PATH)

df.head()
```

```
# Общая информация о данных
df.info()
```

```
# Базовые статистики по числовым признакам
df.describe().T
```

```
# Проверка пропусков
df.isna().mean().sort_values(ascending=False)
```

```
# Гистограммы для числовых признаков
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns.tolist()

if numeric_cols:
    df[numeric_cols].hist(bins=30, figsize=(15, 8))
    plt.tight_layout()
    plt.show()
```

```
# Простая очистка: удалим полностью пустые столбцы и строки с пропусками
df_clean = df.dropna(axis=1, how="all")
df_clean = df_clean.dropna(axis=0, how="any")

df_clean.info()
```

```
# Сохранение очищенного датасета
PROCESSED_DATA_PATH.parent.mkdir(parents=True, exist_ok=True)
df_clean.to_csv(PROCESSED_DATA_PATH, index=False)

PROCESSED_DATA_PATH
```
